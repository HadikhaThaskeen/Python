```python
class Course:
    def __init__(self,course_code,course_name,credit_hours):
        self.course_code = str(course_code)
        self.course_name = str(course_name)
        self.credit_hours = int(credit_hours)
class CoreCourse(Course):
    def __init__(self,course_code,course_name,credit_hours,required_for_major):
        super().__init__(course_code,course_name,credit_hours)
        self.required_for_major = required_for_major
class ElectiveCourse(Course):
    def __init__(self,course_code,course_name,credit_hours,elective_type):
        super().__init__(course_code,cource_name,credit_hours)   
        self.elective_type = elective_type
```

```python
# employee.py
class Employee:
    def __init__(self,name,salary):
         self.name = name
         self.salary = salary
    def get_name(self):
        return f'The name of employee is {self.name}'
    def get_salary(self):
        return f'The salary of the employee is {self.salary}'
    
```
