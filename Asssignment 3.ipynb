```python
#2)A certain cinema currently sells tickets for a full price of 6 pounds, but always sells tickets for half price to people who are less than 
# 16 years old, and for a third of the price for people who are 60 years old or more.
full_price = 6.0
age = int(input('Enter your age'))
if age<16:
    print('Ticket price is',full_price/2)
elif age>=60:
    print('Ticket price is',full_price/3)
else:
    print('Ticket price is',full_price)
```

```python
#3)Write a program to calculate your BMI and give weight status.
weight = float(input('Enter your weight in (kg):'))
height = float(input('Enter your height in (m):'))
BMI = weight/height
if BMI<18.5:
    print('You are underweight')
elif  18.5<=BMI<24.9:
    print('You are normal')
elif 25<=BMI<29.9:
    print('You are overweight')
else:
    print('You are obese')
```

```python
#5 Find the factorial of a given number using loops(note the number is received from the user)
num=int(input('Enter a number'))
factorial =1
if num<0:
    print('Factorial is not defined for negative numbers')
elif num==0 or num==1:
    print(f'The factorial of the {num} is 1.')
else:
    for i in range(1,num+1):
        factorial*=i
    print(f'The factorial of {num} is {factorial}')
```

```python
#6)  Reverse a number using while loop Exercise
num=int(input('enter a number :'))
reverse = 0
original = num
while num>0:
    digit = num%10
    reverse = (reverse*10)+digit
    num =num//10
print('The reverse of',original,'is',reverse)
```

```python
#7)Finding the multiples of a number using loop Exercise 
num= int(input('enter a number'))
limit = int(input('enter a limit'))
print(f'Multiples of {num} up to {limit}:')
for i in range(1,limit+1):
    print(num*i)
```
