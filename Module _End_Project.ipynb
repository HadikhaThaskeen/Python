# Age Group Distribution (bar chart)

```python
df['Age'] = df['Age'].astype(str).str.strip()
df = df[df['Age'].str.isdigit()]
df['Age'] = pd.to_numeric(df['Age'])

df['Age'].head()
```

```python
bins = [18, 25, 35, 45]
labels = ['18-24', '25-34', '35-44']
df['Age Group'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)

age_group_counts = df['Age Group'].value_counts().sort_index()

age_group_counts.plot(kind='bar', color='#301192')
plt.title('Age Group Distribution')
plt.xlabel('Age Group')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.show()
```

# which team and position have the highest salary expenditure

```python
# Salary expenditure by Team
team_salary = df.groupby('Team')['Salary'].sum().sort_values()
team_salary.plot(kind='barh', color='teal')
plt.title('Salary Expenditure by Team')
plt.xlabel('Total Salary')
plt.ylabel('Team')
plt.tight_layout()
plt.show()
```

```python
# Salary expenditure by Position
position_salary = df.groupby('Position')['Salary'].sum().sort_values()
position_salary.plot(kind='barh', color='darkred')
plt.title('Salary Expenditure by Position')
plt.xlabel('Total Salary')
plt.ylabel('Position')
plt.tight_layout()
plt.show()
```
