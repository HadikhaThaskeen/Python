```python
#4)
def convert_to_integer():
    user_input = input('please enter a string:')
    try:
        converted_number = int(user_input)
        print(f"Conversion successful! The integer is: {converted_number}")
    except ValueError:
        print("error: the provided string is not a valid integer.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
```

```python
convert_to_integer()
```

```python
#5)
def positive_integer():
    user_input = input("Enter integers separated by commas: ")
    numbers = [int(item.strip()) for item in user_input.split(",")] # Convert input string into a list of integers
    
    if any(num < 0 for num in numbers):  # Check if any integer is negative
        raise ValueError("Negative integer detected in the list!")
    else:
        print("All integers are non-negative!")
```

```python
positive_integer()
```

```python
#6)
def average_num():
    try:
        user_input = input("Enter a list of integers separated by spaces: ")
        numbers = [int(x) for x in user_input.split()] # convert the string into integer
        average = sum(numbers) / len(numbers)
        print(f"The average of the given integers is: {average}")
    except ValueError:
        print("Error: Please ensure all inputs are valid integers.")
    except ZeroDivisionError:
        print("Error: No integers were provided.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    finally:
        print("Program has finished running.")
```

```python
average_num()
```
